## Linked List Oprations

```python
class Node :
    def __init__(self , data):
        self.data = data 
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None
        self.count = 0

    def printList(self):
        current = self.head
        while current :
            print(current.data , end=' ')
            current = current.next

    def insertAtStart(self , data):
        newNode = Node(data)
        if self.head == None :
            self.head = newNode
        else :
            newNode.next = self.head
            self.head = newNode
        self.count +=1

    def inserAtEnd (self , data):
        newNode = Node(data)
        prevNode = self.head
        if self.head == None :
            self.head = newNode
        else :
            while prevNode.next != None :
                prevNode = prevNode.next
            prevNode.next = newNode
        self.count +=1
        
    def insertBetween(self , previousNode , data):
        if previousNode == None:
            print('Not Found')
            return
        else:
            newNode = Node(data)
            newNode.next = previousNode.next
            previousNode.next = newNode
        self.count +=1

    
    def deleteItem(self, key):
        temp = self.head
        if temp.next :
            if temp.data == key:
                self.head = temp.next
                temp = None
            else :
                while temp.next != None:
                    if temp.data == key:
                        break
                    prev = temp
                    temp = temp.next
                if temp == None:
                    print('Not Found')
                prev.next = temp.next
                return
        self.count -=1
        
    def deleteFirst(self):
        temp = self.head
        self.head = temp.next
        temp = None

    def deleteSimple(self):
        self.head = self.head.next
    
        

list = LinkedList()
list.insertAtStart(5)
list.insertAtStart(4)
list.insertAtStart(3)
list.inserAtEnd(6)
list.insertBetween(list.head,99)
list.printList()
list.deleteItem(99)
list.printList()
list.count
```
