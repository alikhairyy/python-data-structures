## Part 1 — Generate the dataset

Each row represents one order. The numeric columns are:

| Column | Meaning |
|---|---|
| `order_id` | Unique order number |
| `product_id` | Product category ID (1–5) |
| `units_sold` | Number of units in the order |
| `unit_price` | Price per unit, in currency units |
| `discount_pct` | Discount percentage (0–30) |
| `customer_rating` | Customer rating (1–5) |
| `delivery_days` | Delivery time in days |

**Data quality issues intentionally included:** some missing values (`NaN`) in ratings and delivery times, plus one invalid negative quantity and one unusually high unit price. You will investigate these later.

```python
import numpy as np

np.random.seed(42)

n_orders = 60

order_id = np.arange(1001, 1001 + n_orders)
product_id = np.random.randint(1, 6, size=n_orders)
units_sold = np.random.randint(1, 11, size=n_orders).astype(float)
unit_price = np.random.choice([50, 75, 100, 125, 150, 200], size=n_orders).astype(float)
discount_pct = np.random.choice([0, 5, 10, 15, 20, 25, 30], size=n_orders).astype(float)
customer_rating = np.random.randint(1, 6, size=n_orders).astype(float)
delivery_days = np.random.randint(1, 8, size=n_orders).astype(float)

# Introduce data-quality issues intentionally.
customer_rating[[4, 17, 33, 48]] = np.nan
delivery_days[[8, 21, 42]] = np.nan
units_sold[12] = -3            # Invalid quantity
unit_price[39] = 9999          # Unusually high price; investigate it

# Combine the columns into one 2D array.
# Column order: order_id, product_id, units_sold, unit_price,
#               discount_pct, customer_rating, delivery_days
data = np.column_stack([
    order_id, product_id, units_sold, unit_price,
    discount_pct, customer_rating, delivery_days
])

print("Dataset generated successfully!")
print("Shape:", data.shape)
print(data[:10])
```
