```python
from collections import deque
from datetime import datetime

# -----------------------------

class Customer:
    def __init__(self, customer_id, name, cart):
        self.customer_id = customer_id
        self.name = name
        self.cart = cart  


class SalesTransactionSystem:
    def __init__(self):
        self.checkout_queue = deque()   # Queue
        self.transaction_stack = []     # Stack
        self.sales_data = {}            # Hash Map

    # Add customer to queue
    def add_customer(self, customer):
        self.checkout_queue.append(customer)
        print(f"{customer.name} added to checkout queue.")

    
    def process_transaction(self):
        if not self.checkout_queue:
            print("No customers in queue.")
            return


        customer = self.checkout_queue.popleft()

        total_amount = sum(customer.cart.values())

      
        for product, price in customer.cart.items():
            if product in self.sales_data:
                self.sales_data[product] += price
            else:
                self.sales_data[product] = price

        
        transaction = {
            "customer_id": customer.customer_id,
            "customer_name": customer.name,
            "total": total_amount,
            "time": datetime.now()
        }

       
        self.transaction_stack.append(transaction)

        print(f"Transaction completed for {customer.name}. Total = ${total_amount}")

    # Refund last transaction
    def refund_last_transaction(self):
        if not self.transaction_stack:
            print("No transactions to refund.")
            return

        transaction = self.transaction_stack.pop()
        print(f"Refund issued for {transaction['customer_name']} (${transaction['total']})")

    # Sales report
    def sales_report(self):
        print("\n--- Sales Report ---")
        for product, revenue in self.sales_data.items():
            print(f"{product}: ${revenue}")
```

```python
system = SalesTransactionSystem()

c1 = Customer(1, "Ali", {"Milk": 5, "Bread": 3})
c2 = Customer(2, "Hana", {"Juice": 4})

system.add_customer(c1)
system.add_customer(c2)

system.process_transaction()
system.process_transaction()

system.refund_last_transaction()
system.sales_report()

```
